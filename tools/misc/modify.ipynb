Under here was used for research assignment to change the amount of lidar point cloud data.

```python
import pickle
import numpy as np

import numpy as np

def calculate_num_points(lidar_path):
    # Load the LiDAR point cloud data from the modified file
    # Example assumes the data is stored in a numpy array
    point_cloud = np.fromfile(lidar_path, dtype=np.float32).reshape(-1, 5)  # Adjust based on your data format
    return point_cloud.shape[0]  # Return the number of points
```

```python
with open('/home/mingdayang/mmdetection3d/data/nuscenes/mmdet3d_bevformer/mini_nuscenes_infos_temporal_val.pkl', 'rb') as f:
    data = pickle.load(f)
```

```python
print(data['infos'][1]['num_lidar_pts'])
```

```python
for item in data['infos']:
    # Update the LiDAR path to point to your modified LiDAR files
    item['lidar_path'] = item['lidar_path'].replace('data/nuscenes/', 'save_root/beam_missing/24beam/')
    new_num_points = calculate_num_points(item['lidar_path'])
    # print(f"old {item['num_lidar_pts']} new: {new_num_points}")
    item['num_lidar_pts'] = new_num_points
```
