```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
# Generate a random image of size 200x200 with 3 RGB channels
random_image = np.random.randint(0, 256, (200, 200, 3), dtype=np.uint8)

# Downscale the image to half of its original size
downscaled_image = cv2.resize(random_image, (0, 0), fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)

# Upscale the image to twice its original size
upscaled_image = cv2.resize(random_image, (0, 0), fx=2.0, fy=2.0, interpolation=cv2.INTER_LINEAR)

# Display the original, downscaled, and upscaled images using matplotlib
fig, axs = plt.subplots(1, 3, figsize=(15, 5))

axs[0].imshow(random_image)
axs[0].set_title("Original Image (200x200)")
axs[0].axis('off')

axs[1].imshow(downscaled_image)
axs[1].set_title("Downscaled Image (100x100)")
axs[1].axis('off')

axs[2].imshow(upscaled_image)
axs[2].set_title("Upscaled Image (400x400)")
axs[2].axis('off')

plt.tight_layout()
plt.show()
```
